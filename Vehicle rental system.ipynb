```python
# VEHICLE RENTAL SYSTEM

print("---------------WELCOME TO VEHICLE RENTAL SYSTEM---------------")
print()
print()

# GETTING CUSTOMER DETAILS

customer_name = input("ENTER YOUR NAME: ")
print()

mobile_no = input("ENTER YOUR MOBILE NUMBER: ")
print()

mob = len(mobile_no)

if mob == 10:
    print("Mobile number is valid")
    print()

    # VEHICLE TYPE DETAILS

    vehicle_type = {
        1: "Bike",
        2: "Car"
    }

    print("----------AVAILABLE VEHICLE TYPES----------")
    for key, value in vehicle_type.items():
        print(key, value)

    print()

    your_vehicle_type = int(input("ENTER YOUR VEHICLE TYPE NUMBER: "))
    print()

    if your_vehicle_type > 2:
        print("Please enter a valid vehicle type!")
    else:

        # VEHICLE DETAILS

        vehicles = {
            1: [
                {
                    "id": 1,
                    "name": "Yamaha FZ",
                    "price": 800,
                    "fuel": "Petrol",
                    "seats": 2
                },
                {
                    "id": 2,
                    "name": "Royal Enfield Classic 350",
                    "price": 1200,
                    "fuel": "Petrol",
                    "seats": 2
                },
                {
                    "id": 3,
                    "name": "KTM Duke 200",
                    "price": 1000,
                    "fuel": "Petrol",
                    "seats": 2
                }
            ],

            2: [
                {
                    "id": 1,
                    "name": "Hyundai i20",
                    "price": 1800,
                    "fuel": "Petrol",
                    "seats": 5
                },
                {
                    "id": 2,
                    "name": "Tata Nexon",
                    "price": 2200,
                    "fuel": "Petrol",
                    "seats": 5
                },
                {
                    "id": 3,
                    "name": "Toyota Innova",
                    "price": 3000,
                    "fuel": "Diesel",
                    "seats": 7
                }
            ]
        }

        print("----------HERE ARE YOUR VEHICLE OPTIONS----------")
        print()

        selected_vehicles = vehicles.get(your_vehicle_type)

        for vehicle in selected_vehicles:
            print(
                f"{vehicle['id']}. {vehicle['name']} | "
                f"₹{vehicle['price']} per day | "
                f"Fuel: {vehicle['fuel']} | "
                f"Seats: {vehicle['seats']}"
            )

        print()

        vehicle_id = int(input("ENTER VEHICLE ID: "))
        print()

        # FIND SELECTED VEHICLE

        for vehicle in selected_vehicles:

            if vehicle["id"] == vehicle_id:

                print("----------YOUR VEHICLE DETAILS----------")
                print()

                print("Customer Name:", customer_name)
                print("Mobile Number:", mobile_no)
                print("Vehicle Type:", vehicle_type.get(your_vehicle_type))
                print("Vehicle Name:", vehicle["name"])
                print("Price Per Day:", "₹", vehicle["price"])
                print("Fuel Type:", vehicle["fuel"])
                print("Seats:", vehicle["seats"])

                print()

                # GET RENTAL DAYS

                days = int(input("ENTER NUMBER OF RENTAL DAYS: "))
                print()

                if days > 0:

                    total = vehicle["price"] * days

                    print("----------RENTAL BILL----------")
                    print()

                    print("Customer Name:", customer_name)
                    print("Mobile Number:", mobile_no)
                    print("Vehicle:", vehicle["name"])
                    print("Price Per Day:", "₹", vehicle["price"])
                    print("Rental Days:", days)
                    print("Total Amount:", "₹", total)

                    print()

                    confirm_rental = input(
                        "CONFIRM RENTAL - CHOOSE Y/N: "
                    )

                    if confirm_rental == "Y":
                        print()
                        print("Rental Successful 🚗")
                        print()
                    else:
                        print()
                        print("Rental Cancelled ❌")
                        print()

                    print("Thank You 🎉")

                else:
                    print("Please enter a valid number of days!")

        if vehicle_id > 3:
            print("Please enter a valid vehicle ID!")

else:
    print("Enter a valid mobile number!")
```
